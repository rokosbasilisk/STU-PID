```python
import os
import shutil
import csv
import PyPDF2
import openai
import pandas as pd
from tqdm import tqdm

# ----------------------------------------
# Configuration - Update these variables
# ----------------------------------------
INPUT_FOLDER = "cot_papers/"        # Folder containing all research paper PDFs
OUTPUT_FOLDER = "cot_papers_selected"   # Folder where relevant PDFs will be moved
CSV_PATH = "cot_papers_selected.csv"            # Path where the CSV results will be saved
OPENAI_API_KEY = os.environ["OPENAI_API_KEY"]           # Replace with your OpenAI API key
MODEL_NAME = "gpt-4.1-2025-04-14"                            # Model to use (e.g., "gpt-4o")—ensure your account has access

# ----------------------------------------
# Ensure output folder exists
# ----------------------------------------
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# ----------------------------------------
# Initialize OpenAI
# ----------------------------------------
openai.api_key = OPENAI_API_KEY

# ----------------------------------------
# Define the detailed prompt template
# ----------------------------------------
PROMPT_TEMPLATE = """
You are an expert research assistant. Your task is to read the abstract (or the first 1000 characters of the paper's content if an abstract section is not explicitly marked) and determine whether the paper is focused on reducing redundancy in chain-of-thought (CoT) reasoning and making CoT more computationally efficient. If the paper is indeed related to improving reasoning efficiency (e.g., introducing benchmarks like Think-Bench, metrics for overthinking, methods to cut down redundant tokens, or novel activation‐steering approaches such as RASPID), respond with EXACTLY three lines, each separated by a newline:

1. A single-word answer: “Yes” if the paper is about CoT efficiency/redundancy reduction, otherwise “No”.
2. A one‐sentence review of why it is or isn't related (e.g., “This paper introduces Think-Bench, a benchmark for evaluating LRM reasoning efficiency and highlights overthinking issues.”).
3. A concise label or keyword list summarizing the core topic (e.g., “Think-Bench; Overthinking; CoT Efficiency” or “Not Relevant; Other Topic”).

Below is the example abstract for reference, demonstrating the style and content you should detect:

---
Example Abstract:
“Large reasoning models (LRMs) have achieved impressive performance in complex tasks, often outperforming conventional large language models (LLMs). However, the prevalent issue of overthinking severely limits their computational efficiency. Overthinking occurs when models generate excessive and redundant tokens that contribute little to accurate outcomes, especially in simple tasks, resulting in a significant waste of computational resources. To systematically investigate this issue, we introduce Think-Bench, a benchmark designed to evaluate the reasoning efficiency of LRMs. We also propose novel efficiency metrics and conduct a comprehensive evaluation of various LRMs across multiple dimensions, including the reasoning process, outcome quality, and chain-of-thought (CoT) characteristics. Our analysis reveals that most LRMs exhibit overthinking in handling easy questions, generating unnecessarily lengthy reasoning chains. While many LRMs demonstrate high CoT quality, several suffer from low efficiency. We hope that Think-Bench can serve as a robust foundation for advancing research into LRMs.”
---

Now evaluate the following text (which may be truncated to the first 1000 characters if needed). Only output the three lines as specified above—do NOT include any additional commentary.
"""

# ----------------------------------------
# Function to extract text (prefer abstract) from a PDF
# ----------------------------------------
def extract_text_from_pdf(pdf_path, max_chars=10000):
    """
    Extracts text from the first 10000 characters of the PDF, 
    attempting to find the 'Abstract' section. If 'Abstract' isn't found,
    returns the first max_chars characters of the entire text.
    """
    reader = PyPDF2.PdfReader(pdf_path)
    full_text = ""
    for page in reader.pages:
        try:
            full_text += page.extract_text() + "\n"
        except Exception:
            continue

    # Try to locate 'Abstract' section
    lowered = full_text.lower()
    if "abstract" in lowered:
        # Find the slice starting at 'abstract'
        idx = lowered.index("abstract")
        # Extract up to next 1000 chars after 'Abstract' heading for context
        return full_text[idx : idx + max_chars]
    else:
        # Return first max_chars characters of entire text
        return full_text[:max_chars]

# ----------------------------------------
# Function to query OpenAI GPT for relevance & review
# ----------------------------------------
def analyze_paper(text_snippet):
    """
    Sends the snippet to OpenAI with the detailed prompt template
    and returns a tuple: (is_relevant (bool), one_line_review (str), labels (str)).
    """
    prompt = PROMPT_TEMPLATE + "\n" + text_snippet.strip() + "\n---\nAnswer:\n"
    try:
        response = openai.ChatCompletion.create(
            model=MODEL_NAME,
            messages=[
                {"role": "system", "content": "You are a helpful research assistant."},
                {"role": "user", "content": prompt}
            ],
            temperature=0.0,
            max_tokens=150,
            top_p=1.0,
            frequency_penalty=0.0,
            presence_penalty=0.0
        )
        output = response.choices[0].message.content.strip().split("\n")
        # Expecting exactly 3 lines: ["Yes"/"No", "One-sentence review.", "Labels"]
        if len(output) >= 3:
            yes_no = output[0].strip()
            review = output[1].strip()
            labels = output[2].strip()
            is_relevant = yes_no.lower() == "yes"
            return is_relevant, review, labels
        else:
            # Fallback if the format isn't as expected
            return False, "FormatError: Unexpected response format", ""
    except Exception as e:
        return False, f"APIError: {str(e)}", ""

# ----------------------------------------
# Main loop: iterate through PDFs, analyze, record results, move files
# ----------------------------------------
results = []  # Will store dicts: {"title": ..., "review": ..., "labels": ...}

for filename in tqdm(os.listdir(INPUT_FOLDER), desc="Processing PDFs"):
    if not filename.lower().endswith(".pdf"):
        continue

    pdf_path = os.path.join(INPUT_FOLDER, filename)
    try:
        # Attempt to read PDF metadata to get the Title; fallback to filename
        reader = PyPDF2.PdfReader(pdf_path)
        title = reader.metadata.title if reader.metadata and reader.metadata.title else filename
    except Exception:
        title = filename

    # Extract the abstract or first chunk of text
    snippet = extract_text_from_pdf(pdf_path, max_chars=2000)

    # Query the model
    is_relevant, one_line_review, labels = analyze_paper(snippet)

    # Append to results
    results.append({
        "title": title,
        "filename": filename,
        "is_relevant": is_relevant,
        "one_line_review": one_line_review,
        "labels": labels
    })

    # If relevant, move the PDF to OUTPUT_FOLDER
    if is_relevant:
        dest_path = os.path.join(OUTPUT_FOLDER, filename)
        shutil.move(pdf_path, dest_path)

# ----------------------------------------
# Save results to CSV
# ----------------------------------------
df = pd.DataFrame(results)
df.to_csv(CSV_PATH, index=False, quoting=csv.QUOTE_MINIMAL)

print(f"Analysis complete. Results saved to '{CSV_PATH}'. Relevant papers moved to '{OUTPUT_FOLDER}'.")
```
